```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow:", tf.__version__)
```

```python
# Input range
x = tf.linspace(-10.0, 10.0, 400)

# Parameters (choose anything)
w = tf.constant(1.2)
b = tf.constant(-0.5)

# Compute y = sigmoid(wx + b)
y = tf.sigmoid(w * x + b)

# Quick check
print("x shape:", x.shape)
print("y shape:", y.shape)
```

```python
with tf.GradientTape() as tape:
    tape.watch(x)                 # tell TF to track x
    y = tf.sigmoid(w * x + b)     # recompute inside the tape context

dy_dx = tape.gradient(y, x)       # derivative of y wrt x

print("dy/dx shape:", dy_dx.shape)
print("dy/dx sample:", dy_dx[:5].numpy())
```

```python
plt.figure(figsize=(9,5))
plt.plot(x.numpy(), y.numpy(), label="y = sigmoid(w*x + b)")
plt.plot(x.numpy(), dy_dx.numpy(), label="dy/dx (gradient)")
plt.xlabel("x")
plt.ylabel("value")
plt.title("Sigmoid function and its gradient")
plt.grid(True)
plt.legend()
plt.show()
```

```python
x_np = x.numpy()
y_np = y.numpy()

# Numerical derivative (finite difference)
dy_dx_num = np.gradient(y_np, x_np)

# Mean absolute error between TF gradient and numerical gradient
mae = np.mean(np.abs(dy_dx.numpy() - dy_dx_num))
print("MAE(TF gradient vs numerical):", mae)

plt.figure(figsize=(9,5))
plt.plot(x_np, dy_dx.numpy(), label="TF gradient dy/dx")
plt.plot(x_np, dy_dx_num, linestyle="--", label="Numerical gradient")
plt.xlabel("x")
plt.ylabel("dy/dx")
plt.title("Gradient check: TensorFlow vs numerical approximation")
plt.grid(True)
plt.legend()
plt.show()
```

The sigmoid function is S-shaped, approaching 0 for large negative inputs and 1 for large positive inputs. Its gradient is largest near the midpoint of the curve, where the function changes most rapidly, and approaches zero in the saturated regions.
The gradient computed using TensorFlow’s automatic differentiation closely matches the numerical gradient approximation, with a mean absolute error of 0.000011626707. This confirms the correctness of the gradient computation.
The visualisation illustrates the vanishing gradient effect: as the sigmoid output saturates, the gradient becomes very small, which can slow learning in deep neural networks.
